# Read from silver table/s

```python
%sql
SELECT 
    ROW_NUMBER() OVER(ORDER BY ci.customer_id) as customer_key
    , ci.customer_id
    , ci.firstname
    , ci.lastname
    , ci.martial_status
    , CASE
        WHEN ci.gender <> 'n/a' THEN ci.gender
        else coalesce(ci.gender, 'n/a')
    END as gender
    , ci.create_date
FROM workspace.silver.crm_customers ci
limit 10
```
